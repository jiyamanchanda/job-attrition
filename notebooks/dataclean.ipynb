```python
# Convert dates into datetime format for investigation
hire_dates = pd.to_datetime(df["hire_date"])
exit_dates = pd.to_datetime(df["exit_date"], errors="coerce")

# Check whether exit dates occur before hire dates
invalid_exit_order = (
    exit_dates.notna() &
    (exit_dates < hire_dates)
)

print("Exit dates before hire dates:", invalid_exit_order.sum())

# Calculate completed years between hire and exit
completed_tenure = (
    exit_dates.dt.year - hire_dates.dt.year
    - (
        (exit_dates.dt.month < hire_dates.dt.month) |
        (
            (exit_dates.dt.month == hire_dates.dt.month) &
            (exit_dates.dt.day < hire_dates.dt.day)
        )
    ).astype(int)
)

# Compare recorded tenure with calculated tenure for employees who left
tenure_comparison = pd.DataFrame({
    "Recorded_Tenure": df.loc[exit_dates.notna(), "years_at_company"],
    "Calculated_Tenure": completed_tenure[exit_dates.notna()]
})

tenure_comparison["Difference"] = (
    tenure_comparison["Recorded_Tenure"]
    - tenure_comparison["Calculated_Tenure"]
)

print("\nTenure comparison for employees who left:")
display(tenure_comparison.describe().T)

print(
    "\nEmployees with matching recorded and calculated tenure:",
    (tenure_comparison["Difference"] == 0).sum()
)

print(
    "Employees with different recorded and calculated tenure:",
    (tenure_comparison["Difference"] != 0).sum()
)

# Show examples of discrepancies, if any
display(
    tenure_comparison[tenure_comparison["Difference"] != 0].head(10)
)
```

```python
# Identify all inconsistent records
inconsistent_records = df.loc[
    tenure_comparison["Difference"].reindex(df.index).notna()
    & (tenure_comparison["Difference"].reindex(df.index) != 0),
    ["employee_id", "hire_date", "exit_date", "years_at_company"]
].copy()

# Add calculated tenure
inconsistent_records["calculated_tenure"] = (
    completed_tenure.loc[inconsistent_records.index]
)

display(inconsistent_records)
```

```python
# Extract hiring year and month for investigation only
hire_year = df_clean["hire_date"].dt.year
hire_month = df_clean["hire_date"].dt.month

print("HIRING DATE RANGE")

print("Earliest hire date:", df_clean["hire_date"].min())
print("Latest hire date:", df_clean["hire_date"].max())

print("\nEMPLOYEES HIRED PER YEAR")

display(
    hire_year.value_counts()
    .sort_index()
    .rename("Employee_Count")
)

print("\nEMPLOYEES HIRED PER MONTH")

display(
    hire_month.value_counts()
    .sort_index()
    .rename("Employee_Count")
)
```

```python
# Create temporary investigation DataFrame
hiring_analysis = pd.DataFrame({
    "hire_year": df_clean["hire_date"].dt.year,
    "attrition": df_clean["attrition"]
})

# Calculate employee counts and attrition rates
yearly_attrition = hiring_analysis.groupby("hire_year").agg(
    total_employees=("attrition", "count"),
    employees_left=("attrition", lambda x: (x == "Yes").sum())
)

# Calculate attrition percentage
yearly_attrition["attrition_rate"] = (
    yearly_attrition["employees_left"]
    / yearly_attrition["total_employees"]
    * 100
)

# Display results
display(yearly_attrition.round(2))
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=yearly_attrition.index,
    y=yearly_attrition["attrition_rate"]
)

plt.title("Attrition Rate by Hiring Year")
plt.xlabel("Hiring Year")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Encode attrition target
df_clean["attrition"] = df_clean["attrition"].map({
    "No": 0,
    "Yes": 1
})

# Encode overtime
df_clean["overtime"] = df_clean["overtime"].astype(int)

print("Binary encoding completed.")
```
