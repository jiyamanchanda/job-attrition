```python
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [10,20,15,30,25]

plt.plot(x,y)

plt.title("Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show
```

```python
import matplotlib.pyplot as plt
subjects=["python","java","c++"]
students=[50,30,20]
plt.pie(students, labels=subjects, autopct='%1.1f%%')
plt.title("Student Enrollment")
plt.show()
```
